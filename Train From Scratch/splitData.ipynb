# Split Data

This Jupyter Notebook is used to split a large sum of labeled data into smaller subsets used for training, testing, and validating a self trained YOLO OBB model. To begin make sure that you have two folders in the unsplit_data directory of this project, one named 'images' and another named 'annotations'. The 'annotation' directory should contain .txt files of YOLO OBB format bounding box coordinates. 

```python
import os
import shutil
import random
```

```python
# Define paths
SOURCE_IMAGES = "/datasets/unsplit_data/images"
SOURCE_LABELS = "/datasets/unsplit_data/annotations"
OUTPUT_DIR = "datasets"
```

```python
# Create train, val, test folders
sets = ["train", "val", "test"]
subdirs = ["images", "labels"]
```

```python
for s in sets:
    for sub in subdirs:
        os.makedirs(os.path.join(OUTPUT_DIR, s, sub), exist_ok=True)
```

```python
# Get all image files
image_files = [f for f in os.listdir(SOURCE_IMAGES) if f.endswith(('.jpg', '.png', '.jpeg'))]
```

```python
# Shuffle and split
random.shuffle(image_files)
num_images = len(image_files)

train_split = int(num_images * 0.8)
val_split = int(num_images * 0.1)

train_files = image_files[:train_split]
val_files = image_files[train_split:train_split + val_split]
test_files = image_files[train_split + val_split:]
```

```python
# Function to move files
def move_files(file_list, split):
    for file in file_list:
        img_src = os.path.join(SOURCE_IMAGES, file)
        label_src = os.path.join(SOURCE_LABELS, file.replace(os.path.splitext(file)[1], ".txt"))

        img_dst = os.path.join(OUTPUT_DIR, split, "images", file)
        label_dst = os.path.join(OUTPUT_DIR, split, "labels", file.replace(os.path.splitext(file)[1], ".txt"))

        if os.path.exists(img_src):
            shutil.copy(img_src, img_dst)
        if os.path.exists(label_src):
            shutil.copy(label_src, label_dst)
```

```python
# Move files to respective folders
move_files(train_files, "train")
move_files(val_files, "val")
move_files(test_files, "test")

print("Dataset successfully organized!")
```
